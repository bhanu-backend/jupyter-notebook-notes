```python
#slicing 
#       0    1  2   3   4    5  6     7
lis = [ 22 ,33,566 ,22 ,44,789,3456,23455]
#      -8   -7 -6   -5  -4  -3  -2   -1
```

```python
print(lis[2::3]) # [566, 789]
print(lis[5::2]) # [789, 23455]
print(lis[5 : : -1]) #[789, 44, 22, 566, 33, 22]
print(lis[ : -2 : 4]) # [22, 44]
print(lis[ : -5 : -2 ]) # [23455, 789]
print(lis[ -5 : : -2]) # [22, 33]
print(lis[ -4 : 6 : -1]) #[]


```
